1. INTRODUÇÃO — REVISÃO BIBLIOGRÁFICA

A transformação digital do setor agroambiental tem ampliado significativamente a quantidade, a diversidade e a complexidade dos dados utilizados para compreender os sistemas produtivos e os ambientes naturais. Dados provenientes de análises de solo, sensores proximais, imagens de satélite, sistemas de posicionamento global, máquinas agrícolas, estações meteorológicas e experimentos de campo podem ser integrados para caracterizar a variabilidade espacial e temporal dos ambientes e subsidiar processos de tomada de decisão (POPPIEL et al., 2019; ROSIN et al., 2023). Nesse contexto, a ciência de dados constitui um campo interdisciplinar que integra estatística, programação, aprendizado de máquina (machine learning), visualização de dados e conhecimento especializado do domínio, com o objetivo de transformar dados em informações úteis para a pesquisa e o manejo agroambiental.

Nas Ciências Agrárias, a aplicação de métodos computacionais assume papel estratégico devido à elevada heterogeneidade dos sistemas produtivos. Propriedades como textura, teor de matéria orgânica, pH, disponibilidade de nutrientes, umidade e carbono do solo apresentam variabilidade espacial. Da mesma forma, a produtividade, o crescimento vegetal e a ocorrência de estresses são influenciados pela interação entre condições climáticas, relevo, características do solo e práticas de manejo. A espectroscopia do solo, associada ao sensoriamento proximal e ao sensoriamento remoto, pode contribuir para a caracterização dessa variabilidade e para a estimativa de atributos físicos, químicos e mineralógicos do solo (MADEIRA NETTO, 2001; TERRA; DEMATTÊ; VISCARRA ROSSEL, 2015; POPPIEL et al., 2019). Entretanto, a aplicação dessas técnicas exige dados de referência de qualidade, procedimentos consistentes de aquisição e estratégias de validação compatíveis com o contexto de aplicação.

Modelos baseados exclusivamente em médias ou em relações lineares simples podem não representar adequadamente determinados padrões presentes nesses sistemas. A integração do conhecimento agronômico com métodos de ciência de dados permite explorar relações lineares e não lineares, identificar padrões e desenvolver modelos preditivos. Contudo, uma previsão estatística não constitui, por si só, um diagnóstico agronômico ou uma recomendação de manejo. Para que os resultados possam subsidiar decisões, é necessário combinar desempenho preditivo, validação independente, interpretação agronômica e análise das condições de aplicação (TERRA; DEMATTÊ; VISCARRA ROSSEL, 2015; POPPIEL; PAIVA; DEMATTÊ, 2022).

O processo de ciência de dados pode ser organizado como um fluxo de trabalho que envolve a definição do problema, a aquisição e organização dos dados, o controle de qualidade, a análise exploratória, o pré-processamento, a engenharia de atributos (feature engineering), a modelagem, a validação, a interpretação e a comunicação dos resultados. Em aplicações de espectroscopia do solo, esse fluxo pode incluir a aquisição dos espectros, o tratamento dos dados espectrais, a seleção ou transformação das variáveis, a calibração dos modelos, a validação interna e externa e a aplicação das previsões a novas amostras ou locais (TERRA; DEMATTÊ; VISCARRA ROSSEL, 2015; POPPIEL; PAIVA; DEMATTÊ, 2022). Cada etapa deve ser documentada, e os procedimentos que aprendem parâmetros a partir dos dados devem ser ajustados exclusivamente no conjunto de treinamento, a fim de evitar vazamento de dados (data leakage).agroambientais, essa etapa pode ser utilizada para investigar a relação entre atributos do solo e produtividade, entre índices de vegetação e biomassa ou entre propriedades espectrais e carbono do solo.

A análise exploratória de dados (EDA, do inglês exploratory data analysis) constitui uma etapa fundamental desse fluxo de trabalho, pois permite compreender a estrutura do conjunto de dados antes da aplicação dos algoritmos de aprendizado de máquina. Nessa etapa, podem ser investigadas as distribuições das variáveis, os valores ausentes, os valores discrepantes, as correlações, a redundância entre atributos e as possíveis relações lineares e não lineares. Em aplicações de espectroscopia do solo, a análise exploratória também pode revelar diferenças entre bibliotecas espectrais, classes de solo, ambientes de formação, instrumentos e regiões de estudo (MADEIRA NETTO, 2001; DEMATTÊ et al., 2019; POPPIEL et al., 2019). Essas diferenças devem ser consideradas na definição da estratégia de amostragem e de validação.

Após a análise exploratória, o pré-processamento transforma os dados brutos em representações adequadas à modelagem. Na espectroscopia do solo, essa etapa pode envolver a correção ou o tratamento de efeitos de espalhamento, a suavização, a derivação, a correção de linha de base, a normalização e a redução de ruído. A seleção de comprimentos de onda e a redução de dimensionalidade podem ser tratadas como etapas complementares, dependendo do objetivo da análise. A escolha dos procedimentos deve considerar a natureza dos espectros, o atributo de interesse, o instrumento utilizado, a escala de aplicação e a estratégia de validação (TERRA; DEMATTÊ; VISCARRA ROSSEL, 2015; POPPIEL; PAIVA; DEMATTÊ, 2022).

Outro componente essencial é a engenharia de atributos, que consiste na criação, transformação ou seleção de variáveis capazes de representar adequadamente o fenômeno estudado. Nas Ciências Agroambientais, essa abordagem pode envolver a extração de informações de bandas espectrais, a construção de índices de vegetação, a combinação de atributos do relevo e a integração de dados espectrais, climáticos, pedológicos e de sensoriamento remoto. Em estudos de solos, a seleção ou criação de atributos espectrais pode contribuir para a predição de propriedades físicas, químicas e mineralógicas, desde que as variáveis sejam justificadas agronomicamente e avaliadas em dados independentes (DEMATTÊ et al., 2019; ROSIN et al., 2023; PELEGRINO et al., 2025).

O aprendizado de máquina constitui uma das ferramentas utilizadas pela ciência de dados para modelar relações entre variáveis. Os problemas podem ser classificados, entre outros, em tarefas de regressão e de classificação. Na espectroscopia do solo, modelos de regressão podem ser empregados para estimar atributos como carbono orgânico, argila, areia, silte e propriedades mineralógicas. Modelos de classificação podem ser utilizados para discriminar classes de solo, ambientes pedológicos ou níveis de degradação, especialmente quando associados a dados espectrais, ambientais e de sensoriamento remoto (BELLINASO; DEMATTÊ; ROMEIRO, 2010; POPPIEL et al., 2019; ROSIN et al., 2023). A escolha entre regressão e classificação deve ser definida a partir da natureza da variável-alvo e da finalidade da análise.

Entre os algoritmos utilizados em problemas com dados tabulares e espectrais destacam-se o Random Forest, o XGBoost e as máquinas de vetores de suporte (SVM). O Random Forest utiliza múltiplas árvores de decisão e apresenta robustez diante de conjuntos com grande número de atributos. O XGBoost é uma implementação de gradient boosting na qual modelos sucessivos procuram corrigir os erros dos anteriores. As máquinas de vetores de suporte podem ser aplicadas tanto à classificação (SVC), por meio de uma fronteira de separação com margem adequada, quanto à regressão (SVR), e exigem escalonamento das variáveis e ajuste de hiperparâmetros. Esses métodos podem representar relações não lineares entre os atributos espectrais e as propriedades do solo. De modo geral, o desempenho dos modelos depende da qualidade dos dados, da representatividade da biblioteca espectral, do pré-processamento, da seleção dos atributos, do ajuste dos hiperparâmetros e da estratégia de validação adotada (TERRA; DEMATTÊ; VISCARRA ROSSEL, 2015; DEMATTÊ et al., 2019; PELEGRINO et al., 2025). Em dados espaciais ou temporais, a validação deve considerar a possível dependência entre amostras próximas ou pertencentes ao mesmo ambiente.

A utilização de diferentes modelos também permite desenvolver modelos combinados, nos quais as previsões de vários algoritmos são integradas. O ensemble por empilhamento (stacking) combina modelos de base e utiliza um meta-modelo para produzir a previsão final. Para evitar vazamento de dados, as previsões utilizadas no treinamento do meta-modelo devem ser obtidas por um procedimento adequado, como previsões fora da amostra geradas dentro do conjunto de treinamento. O desempenho do ensemble deve ser comparado ao dos modelos individuais em um conjunto de teste independente.
O desempenho de um modelo não deve ser avaliado apenas pela capacidade de produzir previsões no conjunto utilizado para seu desenvolvimento. Em estudos de espectroscopia do solo, também devem ser considerados a representatividade das amostras, a abrangência da biblioteca espectral, a qualidade das análises laboratoriais de referência, a estabilidade da calibração, a incerteza das previsões e a capacidade de aplicação a novos instrumentos, regiões, classes de solo e condições ambientais (TERRA; DEMATTÊ; VISCARRA ROSSEL, 2015; POPPIEL; PAIVA; DEMATTÊ, 2022). Para regressão, métricas como R², RMSE (raiz do erro quadrático médio), MAE (erro absoluto médio) e viés permitem avaliar diferentes aspectos do erro preditivo, desde que suas limitações sejam discutidas. Para classificação, podem ser utilizadas acurácia, precisão, sensibilidade (recall), F1-score, área sob a curva ROC (AUC-ROC) e matriz de confusão; em dados desbalanceados, convém considerar também a área sob a curva precisão–sensibilidade (AUC-PR).
Outro aspecto fundamental é a interpretabilidade dos modelos. Técnicas de inteligência artificial explicável (XAI), como o SHAP, permitem analisar a contribuição das variáveis para previsões globais e individuais. Na espectroscopia do solo, essa análise pode auxiliar na formulação de hipóteses sobre regiões espectrais associadas a determinados atributos, grupos funcionais ou minerais. Contudo, a contribuição preditiva indicada pelo SHAP descreve o comportamento do modelo e não constitui, isoladamente, evidência de relação causal nem identificação espectroscópica definitiva. A interpretação deve ser confrontada com conhecimento espectroscópico, informações laboratoriais e avaliações da estabilidade das explicações (DEMATTÊ et al., 2019; ROSIN et al., 2023).

A capacidade de generalização dos modelos representa outro desafio central. Um modelo desenvolvido com dados de determinada região, safra, classe de solo, instrumento ou condição ambiental pode apresentar desempenho diferente quando aplicado a outro conjunto de amostras. Essa redução de desempenho pode estar associada à mudança na distribuição dos dados (dataset shift), que ocorre quando a distribuição dos dados no contexto de aplicação ou de teste difere daquela dos dados utilizados para o treinamento. Também podem ocorrer mudanças na relação entre os atributos e a variável-alvo. Por isso, é necessário distinguir a generalização para dados não observados, mas pertencentes a uma população semelhante, da transferência do modelo para um novo contexto (POPPIEL; PAIVA; DEMATTÊ, 2022).

A implementação rotineira da espectroscopia do solo requer validação independente, avaliação da transferência para novos contextos e, quando necessário, recalibração do modelo. A recalibração deve ser realizada com amostras de referência representativas do ambiente, instrumento ou população em que o modelo será utilizado. Em dados agroambientais, estratégias de validação espacial, temporal, por região, por safra ou por instrumento podem ser mais informativas do que uma divisão aleatória simples (POPPIEL; PAIVA; DEMATTÊ, 2022).

Assim, a ciência de dados aplicada às Ciências Agroambientais deve ser compreendida não apenas como a aplicação de algoritmos de aprendizado de máquina, mas como um fluxo de trabalho científico completo de transformação de dados em conhecimento. No caso dos solos, a integração entre análises laboratoriais tradicionais, espectroscopia, sensoriamento proximal, sensoriamento remoto e modelagem estatística pode ampliar a capacidade de caracterização, classificação e mapeamento dos ambientes pedológicos (MADEIRA NETTO, 2001; POPPIEL et al., 2019; POPPIEL; PAIVA; DEMATTÊ, 2022). Os resultados preditivos podem subsidiar diagnósticos e recomendações de manejo quando forem apoiados por validação adequada, interpretação agronômica, análise de incertezas e avaliação das condições de transferência.

3. FUNDAMENTOS DE CIÊNCIA DE DADOS

# FUNDAMENTOS DE CIÊNCIA DE DADOS APLICADA À ANÁLISE AGROAMBIENTAL

A ciência de dados aplicada ao contexto agroambiental consiste na integração entre dados espectrais, laboratoriais e espaciais; estatística; programação; aprendizado de máquina; conhecimento de solos; e critérios agronômicos. Seu objetivo é transformar dados brutos em informações confiáveis, conhecimento interpretável e subsídios para a tomada de decisão.

Nesse contexto, o aprendizado de máquina representa apenas uma parte do processo. Ele reúne métodos utilizados em etapas específicas, como previsão de atributos do solo, classificação de condições agroambientais e detecção de amostras anômalas. Portanto, aprendizado de máquina não é sinônimo de ciência de dados, mas uma ferramenta inserida em um fluxo metodológico mais amplo.

No contexto agroambiental, o processo pode ser representado por:

> **dados → informação → conhecimento → diagnóstico → recomendação → decisão agronômica**

Os dados podem incluir espectros MIR e NIR, resultados de análises laboratoriais, coordenadas geográficas, propriedades físicas e químicas do solo, informações ambientais e observações de campo. Após a organização, o controle de qualidade e o processamento adequado, esses dados podem ser utilizados para caracterizar a variabilidade do solo, identificar padrões espectrais, estimar atributos e reconhecer possíveis limitações agroambientais.

Ao longo deste material, os termos a seguir são utilizados com significados específicos:

- **Fluxo de trabalho:** sequência metodológica completa que envolve a definição do problema agroambiental, a aquisição e organização dos dados, o controle de qualidade, a análise exploratória, o pré-processamento, a engenharia de atributos, a modelagem, a validação e a interpretação dos resultados.

- **Modelo preditivo:** algoritmo treinado com dados de referência para estimar uma variável-alvo ou classificar uma condição agroambiental. Exemplos incluem a previsão de carbono do solo, a classificação de acidez ou compactação e a identificação de possíveis deficiências nutricionais.

- **Aprendizado de máquina:** conjunto de métodos utilizados para construir modelos de regressão, classificação ou detecção de anomalias. Esses métodos são empregados dentro do fluxo de trabalho, mas não substituem as etapas de aquisição, controle de qualidade, interpretação agronômica e validação em campo.

- **Interpretação:** análise do comportamento do modelo, de seu desempenho, de suas limitações e das contribuições das variáveis ou bandas espectrais para as previsões. A interpretação também verifica se os resultados são coerentes com o conhecimento espectroscópico e agronômico.

- **Diagnóstico agroambiental:** caracterização da variabilidade espacial e das condições do solo, considerando os resultados dos modelos, os dados laboratoriais, os padrões espectrais, os fatores ambientais e a incerteza das estimativas.

- **Recomendação de manejo:** definição de possíveis ações agronômicas a partir do diagnóstico, do conhecimento técnico, dos limites de confiança do modelo, dos custos, das restrições operacionais e da necessidade de confirmação por análises laboratoriais ou observações de campo.

- **Decisão agronômica:** escolha da ação a ser adotada pelo responsável técnico ou produtor, considerando as evidências fornecidas pelo modelo, o diagnóstico, a recomendação de manejo, os riscos e as condições reais da área.

- **Sistema de apoio à decisão:** produto final que integra dados, modelos preditivos, mapas, indicadores de incerteza, diagnósticos e recomendações de manejo para apoiar decisões agronômicas mais eficientes e rastreáveis.

A aplicação da ciência de dados ao solo, portanto, não termina na geração de uma previsão ou na produção de um mapa. Os resultados devem ser interpretados, convertidos em diagnóstico, avaliados quanto à sua confiabilidade e, quando necessário, confirmados em campo. Essa validação permite corrigir inconsistências, atualizar os modelos e retroalimentar o fluxo de trabalho.

Assim, o processo completo pode ser resumido como:

> **dados agroambientais → processamento e modelagem → interpretação → diagnóstico → recomendação de manejo → decisão agronômica → validação em campo**

3.1 Ciclo da ciência de dados

Fluxo de trabalho agroambiental

O fluxo recomendado conecta a pergunta agronômica aos produtos analíticos e às decisões de campo:

```text
problema agroambiental
  -> aquisição dos dados
  -> organização
  -> controle de qualidade
  -> EDA
  -> pré-processamento
  -> atributos espectrais
  -> machine learning
  -> interpretação
  -> diagnóstico
  -> visualização de prioridades
```

### Etapas e entregáveis

1. **Problema agroambiental**: defina o alvo, a unidade de decisão, o horizonte temporal, a população de interesse e o critério de sucesso.
2. **Aquisição dos dados**: reúna dados de solo, colheita, sensores, imagens, clima e topografia, registrando fonte, data, localização, unidades e  limitações.
3. **Organização**: padronize identificadores, formatos, coordenadas, datas e chaves de integração; preserve os dados brutos.
4. **Controle de qualidade**: verifique duplicatas, ausências, unidades, limites plausíveis, valores discrepantes, resolução e consistência espacial
   e temporal.
5. **EDA**: descreva distribuições, variabilidade, correlações, colinearidade, relações não lineares e padrões espaciais ou temporais.
6. **Pré-processamento**: aplique imputação, filtros, escalonamento, normalização, codificação e seleção dentro do treino ou de cada fold.
7. **Atributos espectrais**: derive bandas, índices, picos, profundidades,  derivadas e resumos temporais com parâmetros documentados.
8. **Machine learning**: treine modelos de regressão ou classificação com validação adequada à unidade espacial e temporal.
9. **Interpretação**: use coeficientes, importância, SHAP, curvas e análise de  sensibilidade para explicar o comportamento do modelo.
10. **Diagnóstico**: examine resíduos, erros por grupo, calibração, incerteza,  extrapolação e estabilidade em áreas ou safras não observadas.
11. **Visualização de prioridades**: gere mapas, rankings e tabelas com  legenda, unidades, data, incerteza, critérios de prioridade e limitações.

Cada etapa deve produzir artefatos rastreáveis e indicar sua origem, transformações, versão e responsável. O produto final deve apoiar a decisão
agronômica sem ocultar incertezas: uma prioridade visualizada no mapa precisa estar ligada ao modelo, ao diagnóstico, ao critério de decisão e, quando
possível, à validação em campo.

## Bibliotecas e estrutura do projeto

A implementação pode usar bibliotecas Python diferentes em cada etapa do fluxo agroambiental. Elas são ferramentas de execução, não substituem a
definição do problema, o controle de qualidade, a validação ou a interpretação agronômica.

### Bibliotecas principais

- **pandas**: leitura, organização, integração e manipulação de tabelas;
- **numpy**: operações numéricas e matrizes espectrais;
- **matplotlib**: curvas espectrais, gráficos diagnósticos e visualizações;
- **scikit-learn**: pré-processamento, seleção de atributos, regressão, classificação, detecção de anomalias e validação;
- **XGBoost**: modelos baseados em árvores para regressão e classificação;
- **SHAP**: interpretação de previsões e contribuição de variáveis e bandas;
- **geopandas**: pontos de amostragem, limites de propriedades e dados vetoriais;
- **rasterio**: leitura, processamento e geração de rasters, como mapas de atributos do solo e superfícies de predição.

A escolha deve considerar o tipo e o volume dos dados, a etapa do fluxo, as dependências disponíveis e o objetivo da análise. Registre versões e
parâmetros relevantes para que os resultados possam ser reproduzidos.

### Estrutura sugerida

Uma organização por estágio ajuda a preservar os dados brutos, separar resultados intermediários e rastrear modelos e produtos finais:

```text data/ raw/ processed/ notebooks/ scripts/ models/ figures/ results/```

#### `data/raw/`

Contém os arquivos originais, preservados sem alterações: espectros MIR, NIR ou VIS-NIR; resultados laboratoriais; coordenadas e informações de campo;
imagens de sensoriamento remoto; e arquivos vetoriais e raster. Essa pasta é a fonte de referência e não deve ser sobrescrita durante o processamento.

#### `data/processed/`

Armazena tabelas integradas, unidades padronizadas, espectros filtrados ou suavizados, dados com valores ausentes tratados, atributos espectrais
derivados e partições de treino, validação e teste. Registre a origem, a transformação aplicada e a versão de cada produto.

#### `notebooks/`

Contém notebooks de exploração, desenvolvimento inicial, testes metodológicos, visualização e documentação. O [Boundary.ipynb](../Boundary.ipynb), por
exemplo, reúne etapas de análise espectral, controle de ruído, detecção de outliers, modelagem e diagnóstico agroambiental. A lógica consolidada deve ser gradualmente transferida para scripts, reduzindo a dependência de execução manual de células.

#### `scripts/`

Contém códigos reutilizáveis para importar e organizar dados, executar pré-processamento, calcular atributos, treinar e validar modelos, gerar mapas
e produzir relatórios. Scripts devem receber parâmetros explícitos e evitar depender de variáveis ocultas de um notebook.

#### `models/`

Armazena modelos treinados, pipelines de pré-processamento, parâmetros, versões dos algoritmos, métricas de validação e arquivos auxiliares de
predição. Cada artefato deve identificar a variável-alvo, os dados usados, a data ou versão do treinamento, as unidades e a estratégia de validação.

#### `figures/`

Armazena curvas espectrais, gráficos de correlação, importância de bandas, gráficos SHAP, matrizes de confusão, gráficos de resíduos e diagramas de
validação. Inclua nos nomes ou metadados a versão do experimento e a data de geração.

#### `results/`

Reúne métricas, previsões por amostra, diagnósticos agroambientais, rankings de prioridade, tabelas de recomendação, relatórios de validação e comparações com dados de campo. Resultados finais devem indicar o modelo, a versão dos dados, o período e a incerteza associada.

No repositório atual, `noise_outputs/`, `xgboost_outputs/`, `figures_output/` e `extracted_figures/` funcionam como diretórios de
resultados especializados. Eles devem seguir os mesmos princípios de rastreabilidade e não substituir a preservação dos dados originais.

Estatística aplicada: processo completo

A estatística aplicada organiza a coleta, a descrição, a comparação e a interpretação de dados agroambientais. Ela deve preceder a modelagem e
continuar durante o diagnóstico, ajudando a separar variabilidade real, incerteza de medição, associação estatística e evidência causal.

### 1. Definição da pergunta e da unidade de análise

Defina a pergunta, a população de interesse, a unidade observacional e o alvo. Uma observação pode ser uma amostra de solo, um pixel, um talhão, uma
medição espectral, uma data ou uma combinação dessas unidades. Registre quais observações são replicatas e quais compartilham local, safra, instrumento ou período, pois essa estrutura determina a análise e a validação.

Estabeleça previamente a hipótese ou o objetivo descritivo/preditivo, as variáveis de interesse, as unidades físicas, o período de referência e o
critério de sucesso. Diferencie uma análise exploratória, que busca padrões, de uma análise confirmatória, que avalia uma hipótese especificada.

### 2. Aquisição, organização e desenho amostral

Integre dados de solo, laboratório, sensores, colheita, clima, topografia e imagens por identificadores e coordenadas verificáveis. Preserve a origem,
data, profundidade, método de medição, instrumento, resolução e unidades. Avalie se a amostragem representa a população-alvo e registre seleção,
estratificação, replicação e possíveis fontes de viés.

Não trate observações espacialmente ou temporalmente próximas como independentes sem justificativa. Quando houver pseudorreplicação, agregue na
unidade apropriada ou use métodos que representem a dependência.

### 3. Controle de qualidade e análise exploratória

Verifique tipos, duplicatas, valores ausentes, limites físicos, outliers, unidades incompatíveis e inconsistências entre fontes. Descreva cada variável com contagem válida, média, mediana, quantis, desvio-padrão, amplitude e, quando útil, coeficiente de variação. Use histogramas, densidades, boxplots, curvas espectrais e mapas para examinar distribuição e heterogeneidade.

Analise assimetria, transformações possíveis, variabilidade dentro e entre talhões, regiões, safras e instrumentos. Examine correlações, colinearidade, relações lineares e não lineares, mas não interprete correlação como causalidade. Uma decisão de remover ou manter um outlier deve considerar o domínio e ser registrada.

### 4. Estatística descritiva e incerteza

Relate estimativas com unidades e medidas de dispersão, preferindo intervalos de confiança ou de previsão quando a inferência exigir. Diferencie:

- **variabilidade**: diferenças entre observações da população;
- **erro de medição**: incerteza introduzida pelo instrumento ou protocolo;
- **incerteza amostral**: variação decorrente das unidades observadas;
- **incerteza preditiva**: incerteza ao prever uma nova observação.

Use bootstrap, erros-padrão ou métodos apropriados ao desenho amostral quando forem necessários. Não apresente um intervalo como garantia fora da faixa, população ou condição ambiental usada para calculá-lo.

### 5. Comparações e relações entre variáveis

Escolha testes e estimadores de acordo com escala, distribuição, independência, tamanho amostral e desenho do estudo. Para comparar grupos,
considere diferenças de média ou mediana, tamanho de efeito e intervalos de confiança, não apenas o valor de `p`. Para relações contínuas, examine
correlação, regressão, resíduos, interações e possíveis confundidores.

Corrija ou controle comparações múltiplas quando várias hipóteses forem avaliadas. Em dados agrupados, use modelos hierárquicos, efeitos de grupo ou
erros-padrão robustos quando apropriado. Relate as limitações e não converta associação observacional em afirmação causal sem desenho causal adequado.

### 6. Preparação para modelagem

Divida os dados respeitando grupos, talhões, regiões ou tempo antes de ajustar transformações. Imputação, padronização, normalização, seleção de variáveis, PCA, cálculo de limiares e ajuste de hiperparâmetros devem usar somente o treino ou o fold de treinamento. O conjunto de teste deve permanecer intocado até a avaliação final.

Escolha regressão para alvos contínuos e classificação para categorias definidas operacionalmente. Use validação cruzada compatível com a estrutura
dos dados e compare modelos com os mesmos folds. Em dados espaciais e temporais, uma divisão aleatória pode gerar vazamento e desempenho
artificialmente alto.

### 7. Avaliação, diagnóstico e comunicação

Para regressão, avalie R², RMSE, MAE, viés, resíduos e incerteza. Para classificação, avalie acurácia, precisão, sensibilidade, F1, AUC-ROC, AUC-PR,
matriz de confusão e calibração quando necessário. Em problemas multiclasse,informe a média macro, micro ou ponderada.

Examine resíduos contra previsões e variáveis relevantes, erros por grupo, calibração, extrapolação, estabilidade e generalização. Apresente tabelas,
gráficos e mapas com unidades, escalas, datas, número de observações, intervalos de incerteza e limitações. Resultados estatísticos devem apoiar a
decisão agroambiental, sem substituir critérios agronômicos, validação de campo ou conhecimento do domínio.

3.5 Pré-processamento

## Preparação dos dados e prevenção de vazamento

Antes de treinar um modelo, audite tipos, unidades, identificadores, valores ausentes, duplicatas e limites plausíveis das variáveis. O tratamento deve ser definido a partir do problema e registrado para que o mesmo procedimento seja reproduzido em validação e em novas previsões.

### Valores ausentes e discrepantes

Identifique se a ausência é estrutural, causada pelo instrumento, por nuvens, por limite de detecção ou por falha de coleta. Compare padrões de ausência entre treino e teste e não preencha automaticamente um valor ausente com zero. Use imputação adequada ao tipo de variável, crie um indicador de ausência quando ele tiver significado e avalie a sensibilidade a diferentes estratégias.

Investigue valores discrepantes com regras de domínio, gráficos e métodos robustos. Um outlier pode ser erro de medição, uma observação válida rara ou
um caso fora da população de interesse. Corrija, remova ou mantenha o registro somente com justificativa documentada; não use o rótulo do teste para decidir isso.

### Escala e variáveis categóricas

Padronize ou normalize quando o algoritmo for sensível à escala, às distâncias ou à regularização, como SVM, k-vizinhos, PCA e regressões penalizadas.
Modelos baseados em árvores geralmente não exigem escalonamento, mas a decisão deve ser consistente entre treino e previsão.

Codifique variáveis categóricas de forma compatível com o modelo. One-hot encoding é apropriado para categorias nominais; categorias ordinais devem
preservar sua ordem somente quando essa ordem for justificada. Trate níveis desconhecidos sem falhar em novas amostras e evite codificações que atribuam uma ordem artificial a classes nominais.

### Ordem correta do treinamento

Separe os dados em treino e teste antes de aprender qualquer transformação. Imputação, padronização, normalização, PCA, seleção de variáveis, codificação dependente dos dados e ajuste de hiperparâmetros devem ser ajustados apenas no conjunto de treinamento. Durante validação cruzada, cada uma dessas etapas deve ser ajustada novamente dentro do fold de treinamento, preferencialmente em um `Pipeline` do scikit-learn.

O conjunto de teste deve permanecer **intocado até a avaliação final**. Não use suas estatísticas para imputar, escolher variáveis, definir limiares,
selecionar componentes ou comparar repetidamente modelos. Depois da decisão do modelo, aplique o pipeline já ajustado ao teste uma única vez e registre as métricas e os procedimentos utilizados.

```python
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
])
categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore")),
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_columns),
    ("categorical", categorical_pipeline, categorical_columns),
])
```

Inclua o estimador depois de `preprocessor` no pipeline final e faça a divisão por grupo, talhão, região ou tempo quando essas estruturas puderem produzir dependência entre observações. Vazamento também pode ocorrer por duplicatas, replicatas, atributos calculados com dados futuros ou informações derivadas do rótulo.

3.6 Engenharia de atributos

A engenharia de atributos deve transformar medições brutas em variáveis com justificativa agronômica, mantendo a rastreabilidade da origem, da unidade e da data. Exemplos úteis incluem:

- razões entre atributos do solo, como Ca/Mg, K/CTC, P/argila e MO/argila,
  quando houver hipótese agronômica para sua utilização;
- precipitação acumulada em janelas definidas antes da data do alvo;
- graus-dia calculados a partir de temperaturas-base documentadas;
- déficit hídrico acumulado, com método de balanço hídrico, período e fontes
  meteorológicas registrados.

### Razões entre atributos

Antes de criar uma razão, confirme que o numerador e o denominador estão em unidades compatíveis e que a interpretação do quociente é válida para a
população estudada. Inspecione a distribuição, assimetria, valores extremos e denominadores próximos de zero. Defina uma regra explícita para valores
indefinidos, sem transformar silenciosamente uma divisão inválida em zero.

Avalie se a razão acrescenta informação além das variáveis originais. Razões podem ser redundantes, amplificar ruído ou dificultar a interpretação dos
coeficientes. Compare modelos com e sem a razão usando validação idêntica e calcule as variáveis somente com dados disponíveis no momento da previsão.

### Variáveis meteorológicas e fenológicas

Para precipitação acumulada, informe a janela temporal (por exemplo, dias anteriores à amostragem), a unidade e o tratamento de dados ausentes. Para
graus-dia, registre a temperatura-base, eventuais limites superior e inferior e o período de acumulação. Para déficit hídrico, documente a evapotranspiração, a capacidade de água disponível, o balanço utilizado e se o déficit foi reiniciado a cada ciclo.

Alinhe essas variáveis com a data e a localização da observação e evite usar chuvas ou temperaturas posteriores ao evento que se pretende prever. Em
validação temporal, recalculá-las dentro de cada janela não deve incorporar informação futura.

### NDVI e NDRE no Sentinel-2

Especifique sempre as bandas, o produto e a resolução usados. Um exemplo é:

```text
NDVI = (B8 - B4) / (B8 + B4)
NDRE = (B8 - B5) / (B8 + B5)
```

Nesse caso, `B8` representa NIR, `B4` vermelho e `B5` Red Edge. Prefira reflectância de superfície e documente se foi usada a banda B8 de 10 m ou
uma banda red edge reamostrada. Nuvens, sombras, efeitos atmosféricos, geometria de aquisição e resolução espacial podem alterar os índices.

Aplique máscaras de qualidade antes do cálculo, trate denominadores próximos de zero e preserve a data da imagem. Ao combinar bandas com resoluções
distintas, faça o resampling de modo explícito e consistente. Índices temporais devem ser agregados com uma regra definida (por exemplo, mediana ou
máximo em uma janela fenológica), sem misturar observações inválidas.

Use regressão quando o alvo for uma quantidade contínua. Exemplos incluem:

O modelo prevê a definição operacional adotada; ele não transforma automaticamente um conceito agronômico amplo em uma verdade universal.

3.8 Random Forest

## Random Forest

O **Random Forest** é um algoritmo de aprendizado de máquina baseado em *ensemble*, isto é, combina as previsões de várias árvores de decisão para produzir uma estimativa mais estável e robusta.

Ele pode ser utilizado tanto em:

- **regressão**, para prever valores contínuos;
- **classificação**, para prever categorias ou classes.

## Como funciona

O algoritmo cria várias árvores de decisão utilizando:

1. diferentes amostras dos dados de treinamento;
2. subconjuntos aleatórios de atributos em cada divisão das árvores;
3. regras de decisão sucessivas para separar ou estimar as observações.

No final, as árvores são combinadas:

```text
Dados de entrada
    ↓
Árvore 1 ─┐
Árvore 2  │
Árvore 3  │
...       ├── combinação das previsões
Árvore n ─┘
    ↓
Previsão final
```

Na classificação, a classe final geralmente é definida por votação entre as árvores. Na regressão, a previsão costuma ser obtida pela média das previsões individuais.

## Aplicações agroambientais

O Random Forest pode ser utilizado para:

### Regressão

- produtividade;
- carbono do solo;
- biomassa;
- teor de argila;
- matéria orgânica;
- pH;
- densidade do solo;
- concentração de nutrientes.

### Classificação

- doença/não doença;
- baixo, médio ou alto risco;
- classe de solo;
- compactado/não compactado;
- deficiência/não deficiência;
- degradação/não degradação, desde que essa classe tenha uma definição operacional clara.

No contexto do `Boundary.ipynb`, o Random Forest pode ser aplicado a atributos espectrais MIR/NIR, variáveis químicas e físicas do solo e indicadores de condições agroambientais.

## Vantagens

O Random Forest apresenta algumas características importantes:

- captura relações não lineares;
- representa interações entre variáveis;
- pode lidar com grande número de atributos;
- exige pouca transformação da distribuição das variáveis;
- é aplicável a dados tabulares;
- tende a ser menos sensível a ruído do que uma árvore isolada;
- pode estimar importância das variáveis;
- permite obter probabilidades de classe em problemas de classificação;
- pode ser utilizado como referência para comparação com outros modelos.

Ele é especialmente útil quando a relação entre as variáveis espectrais, climáticas, espaciais e agronômicas não é adequadamente representada por um modelo linear simples.

## Limitações e cuidados

Apesar de robusto, o Random Forest possui limitações:

- pode gerar modelos difíceis de interpretar diretamente;
- pode apresentar importância de variáveis enviesada quando há atributos correlacionados;
- não representa automaticamente dependência espacial ou temporal;
- pode memorizar padrões específicos de uma região quando a validação é inadequada;
- pode apresentar desempenho ruim em classes pouco representadas;
- não deve ser avaliado somente com acurácia ou R²;
- não elimina a necessidade de pré-processamento e validação;
- previsões fora da faixa observada devem ser tratadas com cautela.

Em dados espectrais, muitas bandas podem ser altamente correlacionadas. Nesse caso, a importância de uma banda isolada pode ser instável, pois várias bandas semelhantes podem dividir entre si a contribuição preditiva.

## Principais parâmetros

Alguns parâmetros importantes são:

- `n_estimators`: número de árvores;
- `max_depth`: profundidade máxima das árvores;
- `max_features`: número de atributos considerados em cada divisão;
- `min_samples_split`: número mínimo de amostras para dividir um nó;
- `min_samples_leaf`: número mínimo de amostras em uma folha;
- `class_weight`: ajuste para classes desbalanceadas;
- `random_state`: controle da reprodutibilidade;
- `n_jobs`: número de processadores utilizados.

A escolha desses parâmetros deve ser realizada com validação cruzada ou busca de hiperparâmetros, utilizando exclusivamente os dados de treinamento.

## Fluxo de aplicação

```text
Dados organizados
    ↓
Separação entre treino, validação e teste
    ↓
Tratamento de valores ausentes
    ↓
Codificação de categorias, quando necessária
    ↓
Definição dos hiperparâmetros
    ↓
Treinamento do Random Forest
    ↓
Validação
    ↓
Avaliação no teste intocado
    ↓
Interpretação das variáveis
    ↓
Diagnóstico agroambiental
```

O tratamento de valores ausentes deve ser incluído em um pipeline quando o algoritmo não puder lidar diretamente com esses valores. Além disso, qualquer imputação ou transformação baseada nos dados deve ser ajustada somente no conjunto de treinamento.

## Avaliação

### Para regressão

Podem ser utilizadas:

- R²;
- MAE;
- RMSE;
- viés;
- erro por faixa de valores;
- análise de resíduos;
- validação espacial ou temporal.

### Para classificação

Podem ser utilizadas:

- matriz de confusão;
- precisão;
- recall;
- especificidade;
- F1-score;
- ROC-AUC;
- PR-AUC;
- calibração das probabilidades;
- desempenho por classe.

A validação deve considerar a estrutura dos dados agroambientais. Se houver amostras próximas espacialmente, replicatas ou observações da mesma propriedade, uma divisão aleatória pode superestimar o desempenho. Nesses casos, pode ser mais apropriado separar os dados por:

- propriedade;
- região;
- campanha;
- área de manejo;
- período de coleta.

## Interpretação

A interpretação pode combinar:

- importância por impureza;
- importância por permutação;
- valores SHAP;
- análise de dependência parcial;
- curvas de resposta;
- comparação com conhecimento espectroscópico e agronômico.

Os valores SHAP, por exemplo, podem ajudar a responder:

> Quais bandas, atributos climáticos ou propriedades do solo contribuíram para aumentar ou reduzir uma determinada previsão?

Entretanto, contribuição estatística não significa necessariamente causalidade agronômica. A interpretação deve considerar colinearidade, escala espacial, qualidade dos dados e coerência com o processo físico ou biológico.

## Posição do Random Forest no fluxo de dados

```text
Dados espectrais, laboratoriais, climáticos e espaciais
    ↓
Controle de qualidade
    ↓
EDA
    ↓
Pré-processamento
    ↓
Engenharia de atributos
    ↓
Random Forest
    ↓
Validação
    ↓
Interpretação
    ↓
Diagnóstico agroambiental
    ↓
Recomendação de manejo
```

Assim, o Random Forest é uma ferramenta de modelagem dentro do fluxo de ciência de dados. Ele não substitui a definição do problema, a qualidade dos dados, a validação, a interpretação agronômica ou a confirmação em campo.

a regularização. Uma taxa de aprendizagem menor normalmente exige mais iterações, mas pode produzir um ajuste mais gradual.

3.10 Ensemble por empilhamento (stacking)

O stacking combina previsões de modelos de base (por exemplo, Random Forest, XGBoost e SVM) com um meta-modelo que produz a previsão final. Ele pode ser usado tanto para regressão de atributos do solo (`C_gkg`, por exemplo) quanto para classificação (`SIC_class` ou outra taxonomia).

Para evitar vazamento de dados, o meta-modelo **não deve ser treinado com previsões produzidas no mesmo conjunto usado para ajustar os modelos de base**. As previsões de treinamento devem ser *out-of-fold* (OOF): em cada fold, os modelos de base são ajustados somente na parte de treino e geram previsões na parte de validação. Depois, os modelos de base são reajustados em todo o conjunto de treino para prever o conjunto de teste. O pré-processamento (imputação, suavização, seleção de bandas e escalonamento) também deve ser ajustado dentro de cada fold, preferencialmente em um `Pipeline` do scikit-learn.

Em dados espectrais, mantenha a separação por amostra (`ID_Unico`) antes da validação e não misture bandas ou medições da mesma amostra entre treino e
teste. Avalie regressão com MAE/RMSE e `R²`; para classificação, use acurácia, F1 e, quando as probabilidades forem relevantes, log-loss ou ROC-AUC. Compare o stacking com cada modelo de base usando exatamente os mesmos folds e o mesmo pré-processamento.

3.12 Dados desbalanceados

SMOTE (*Synthetic Minority Over-sampling Technique*) é uma técnica para problemas de classificação com classes desbalanceadas. Em vez de apenas
duplicar observações, ela cria exemplos sintéticos da classe minoritária a partir de combinações entre vizinhos próximos. Isso pode ajudar o classificador a aprender a região da classe minoritária, mas não corrige rótulos incorretos, sobreposição entre classes ou uma amostra originalmente pouco representativa.

O SMOTE deve ser aplicado **somente aos dados de treinamento**. Aplicá-lo antes da divisão treino/teste, ou antes da validação cruzada, permite que exemplos sintéticos incorporem informação das observações de validação ou teste e produz uma estimativa otimista do desempenho. Durante a validação cruzada, use `imblearn.pipeline.Pipeline` para que cada fold faça o ajuste e a sobreamostragem independentemente:

```python
from imblearn.over_sampling import SMOTE
from imblearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer
from sklearn.model_selection import StratifiedKFold, cross_validate
from sklearn.preprocessing import StandardScaler

model = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("smote", SMOTE(random_state=42)),
    ("classifier", RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        n_jobs=-1,
    )),
])

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scores = cross_validate(model, X_train, y_train, cv=cv, scoring=["f1_macro"])
model.fit(X_train, y_train)
predictions = model.predict(X_test)
```

Como o SMOTE utiliza distâncias, imputação e escalonamento devem precedê-lo quando forem necessários. Ajuste os hiperparâmetros (`k_neighbors`,
`sampling_strategy`) usando somente os folds de treinamento e confirme que há amostras minoritárias suficientes para o número de vizinhos escolhido.
Compare o modelo com SMOTE a alternativas como pesos de classe, subamostragem ou classificadores com tratamento nativo de desbalanceamento. Avalie
acurácia junto com métricas menos sensíveis à classe majoritária, como precisão, revocação, F1 por classe, balanced accuracy, matriz de confusão e,
quando apropriado, PR-AUC ou ROC-AUC.

Em dados espectrais, preserve a separação por `ID_Unico` antes da validação e não gere amostras sintéticas que misturem medições relacionadas entre treino e teste. A distribuição das classes deve ser verificada em cada partição, e o conjunto de teste deve permanecer com sua distribuição original para refletir o cenário de uso.

3.13 Avaliação

As métricas devem ser escolhidas de acordo com o tipo de problema e com o custo dos erros. Para regressão:

- **R²**: proporção da variabilidade da resposta explicada pelo modelo. Deve ser interpretado junto com a escala dos erros e pode ser negativo fora da
  amostra.
- **RMSE**: raiz do erro quadrático médio; penaliza mais fortemente erros grandes e mantém a unidade da variável-alvo.
- **MAE**: erro absoluto médio; é mais fácil de interpretar e menos sensível a valores extremos que o RMSE.
- **Viés**: erro médio assinado, por exemplo `mean(y_pred - y_true)`. Indica tendência sistemática de superestimação ou subestimação, que pode permanecer escondida em métricas absolutas.

Para classificação, informe as métricas juntamente com a definição da classe positiva:

- **Acurácia**: proporção de previsões corretas; pode ser enganosa com classes desbalanceadas.
- **Precisão** (*precision*): entre os positivos previstos, proporção correta.
- **Sensibilidade** (*recall*): entre os positivos reais, proporção detectada.
- **F1-score**: média harmônica entre precisão e sensibilidade.
- **AUC-ROC**: capacidade de ordenar positivos acima de negativos em diferentes limiares.
- **Matriz de confusão**: mostra os acertos e tipos de erro por classe.

Em dados desbalanceados, inclua a **AUC-PR** (área sob a curva precisão--sensibilidade), pois ela concentra a avaliação no desempenho da
classe positiva e pode ser mais informativa que a AUC-ROC. Em problemas multiclasse, especifique a estratégia de agregação:

- **Macro**: calcula a métrica por classe e dá o mesmo peso a todas;
- **Micro**: agrega as decisões de todas as classes antes de calcular;
- **Ponderada** (*weighted*): calcula por classe e pondera pelo número de exemplos de cada uma.

O melhor modelo não deve ser escolhido por uma única métrica. Compare desempenho médio e dispersão entre os folds, estabilidade diante de novas
amostras, resíduos e seus padrões, incerteza das previsões, capacidade de generalização em um conjunto de teste intocado e interpretabilidade. Em
regressão, examine gráficos de resíduos contra valores ajustados e variáveis relevantes; em classificação, examine a matriz de confusão por classe e a
calibração das probabilidades quando decisões dependem delas. Registre os folds, intervalos de confiança ou desvios-padrão e o limiar de decisão usado, para que a comparação seja reproduzível.

várias safras e datas, evitando decisões baseadas em um único mapa.

Valide em campo as zonas propostas com amostragem direcionada ou ensaios comparativos. Registre se a divisão é operacionalmente viável e se a
intervenção produz o benefício esperado. Atualize o modelo e as zonas quando novos dados de colheita, solo, imagens ou topografia evidenciarem mudança de cenário.


3.18 Validação e transferência

Avalie os modelos em diferentes áreas, safras e condições ambientais para verificar se o desempenho permanece consistente. **Generalização** é o
desempenho em dados não observados de uma população compatível com a população usada no treinamento. **Transferência** é a aplicação em um contexto
diferente, como outra região, safra, sensor, cultura ou condição ambiental. Transferência não deve ser tratada como uma simples extensão da generalização.

Em uma avaliação de generalização, reserve grupos representativos do mesmo domínio para validação e teste. Em uma avaliação de transferência, use dados independentes do novo contexto, verifique a mudança de distribuição dos preditores e dos rótulos e considere recalibração com amostras de referência representativas. Documente quais dados foram usados na recalibração e não use as mesmas amostras para calibrar e declarar o desempenho final.

### Escolha da validação

Selecione a estratégia de divisão de acordo com a unidade de dependência:

- **Por grupo ou talhão**: mantém medições da mesma unidade na mesma partição;
- **Por região ou área**: testa a transferência espacial para locais não vistos;
- **Por safra ou período**: avalia aplicação em uma temporada ou janela futura;
- **Espacial**: usa blocos ou grupos espaciais para reduzir a influência daautocorrelação entre observações próximas;
- **Temporal**: preserva a ordem cronológica e evita usar o futuro para prevero passado.

Uma divisão aleatória por pixel, ponto ou espectro pode causar vazamento quando observações próximas, replicatas ou medições do mesmo talhão aparecem
simultaneamente no treino e no teste. O desempenho resultante pode refletir semelhança espacial ou temporal, e não capacidade de previsão em novas áreas ou safras.

Use grupos, blocos ou janelas temporais explícitos no procedimento de validação e ajuste o pré-processamento dentro de cada fold. Compare a média e
a dispersão das métricas entre grupos, examine resíduos por área e safra e relate o tamanho e a representatividade de cada partição. Para transferência, inclua validação externa, análise de extrapolação e, quando necessário, recalibração ou atualização do modelo antes do uso operacional.

## Fluxo de trabalho completo: automação e rastreabilidade

O fluxo completo transforma uma pergunta agroambiental em um produto reprodutível para análise ou decisão. A execução deve ser automatizada sempre
que possível, mas cada etapa precisa manter seus parâmetros, entradas, saídas, versões e critérios de validação.

```text
pergunta e objetivo
  -> dados brutos
  -> organização e integração
  -> controle de qualidade
  -> EDA
  -> divisão de treino/validação/teste
  -> pipeline de pré-processamento
  -> engenharia de atributos
  -> treinamento e seleção do modelo
  -> avaliação independente
  -> interpretação e diagnóstico
  -> mapas, prioridades e relatório
```

### Execução passo a passo

1. **Defina o problema**: informe o alvo, a unidade de predição, o horizonte temporal, a população de interesse, as restrições operacionais e o critério  de sucesso.
2. **Registre as fontes**: catalogue arquivos de solo, laboratório, colheita, sensores, Sentinel-2, clima e topografia com identificadores, datas,
   coordenadas, unidades e condições de aquisição.
3. **Integre os dados**: harmonize chaves, projeções, resoluções, profundidade, datas e unidades. Preserve os arquivos originais e gere tabelas ou rasters intermediários versionados.
4. **Faça o controle de qualidade**: verifique duplicatas, ausências, discrepâncias, limites físicos, cobertura de nuvens, saturação, alinhamento
   espacial e consistência temporal. Registre as exclusões e correções.
5. **Realize a EDA**: examine distribuições, variabilidade, correlação, colinearidade, relações não lineares, autocorrelação e diferenças entre
   talhões, regiões e safras.
6. **Defina a validação antes do ajuste**: separe por grupo, talhão, região, bloco espacial ou período conforme a generalização esperada. Reserve o teste final e não o consulte durante decisões de modelagem.
7. **Construa o pipeline**: coloque imputação, codificação, escalonamento, suavização, PCA, seleção de atributos e, quando aplicável, SMOTE dentro do
   treinamento ou de cada fold.
8. **Gere atributos**: calcule bandas, índices, razões, picos, derivadas, agregações temporais e variáveis meteorológicas somente com dados
   disponíveis no momento da previsão.
9. **Treine e compare modelos**: avalie referências simples, PLSR, regressão regularizada, Random Forest, Gradient Boosting, XGBoost ou classificadores adequados. Use os mesmos folds e registre hiperparâmetros.
10. **Selecione com múltiplos critérios**: considere desempenho, dispersão entre  folds, viés, resíduos, incerteza, interpretabilidade, custo e estabilidade  em áreas ou safras não observadas.
11. **Avalie independentemente**: aplique o pipeline final ao teste intocado e relate as métricas, intervalos, matriz de confusão ou gráficos de resíduos. Para transferência, use dados independentes do novo contexto e considere recalibração.
12. **Interprete e diagnostique**: use SHAP, importância, coeficientes, calibração, análise de resíduos e mapas de erro. Verifique extrapolação e
    padrões sistemáticos por grupo.
13. **Produza os resultados**: gere mapas, tabelas, rankings de prioridade e  relatórios com unidades, legenda, data, resolução, incerteza, versão do
    modelo e limitações.
14. **Valide o uso**: compare as previsões com observações de campo e confirme se as prioridades são agronomicamente e operacionalmente viáveis antes de transformá-las em recomendação.

### Automação e reprodutibilidade

Prefira scripts parametrizados para as etapas repetitivas e use notebooks para exploração, visualização e comunicação. Um script de produção deve declarar entradas, parâmetros, ambiente, diretórios de saída e regras de falha; não deve depender de variáveis ocultas ou da ordem manual de execução de células.

Para cada execução, salve um manifesto com a versão do código, data, fontes, filtros, transformações, grupos de validação, sementes aleatórias, métricas e artefatos gerados. Mantenha modelos e pipelines juntos, pois carregar somente o estimador sem o pré-processamento pode produzir previsões incompatíveis.

### Critérios de conclusão

Considere o fluxo concluído somente quando:

- os dados e transformações forem rastreáveis;
- não houver vazamento entre treino, validação e teste;
- a estratégia de validação refletir o uso pretendido;
- as métricas e incertezas estiverem documentadas;
- os erros e limitações tiverem sido diagnosticados;
- os produtos puderem ser reproduzidos por um script ou procedimento
  documentado;
- a interpretação estatística for compatível com a evidência disponível;
- mapas e prioridades tiverem validação agronômica ou de campo quando
  necessários.

## Estatística aplicada: processo completo

A estatística aplicada organiza a coleta, a descrição, a comparação e a interpretação de dados agroambientais. Ela deve preceder a modelagem e
continuar durante o diagnóstico, ajudando a separar variabilidade real, incerteza de medição, associação estatística e evidência causal.

### 1. Definição da pergunta e da unidade de análise

Defina a pergunta, a população de interesse, a unidade observacional e o alvo. Uma observação pode ser uma amostra de solo, um pixel, um talhão, uma
medição espectral, uma data ou uma combinação dessas unidades. Registre quais observações são replicatas e quais compartilham local, safra, instrumento ou período, pois essa estrutura determina a análise e a validação.

Estabeleça previamente a hipótese ou o objetivo descritivo/preditivo, as variáveis de interesse, as unidades físicas, o período de referência e o
critério de sucesso. Diferencie uma análise exploratória, que busca padrões, de uma análise confirmatória, que avalia uma hipótese especificada.

### 2. Aquisição, organização e desenho amostral

Integre dados de solo, laboratório, sensores, colheita, clima, topografia e imagens por identificadores e coordenadas verificáveis. Preserve a origem,
data, profundidade, método de medição, instrumento, resolução e unidades. Avalie se a amostragem representa a população-alvo e registre seleção,
estratificação, replicação e possíveis fontes de viés.

Não trate observações espacialmente ou temporalmente próximas como independentes sem justificativa. Quando houver pseudorreplicação, agregue na
unidade apropriada ou use métodos que representem a dependência.

### 3. Controle de qualidade e análise exploratória

Verifique tipos, duplicatas, valores ausentes, limites físicos, outliers, unidades incompatíveis e inconsistências entre fontes. Descreva cada variável com contagem válida, média, mediana, quantis, desvio-padrão, amplitude e, quando útil, coeficiente de variação. Use histogramas, densidades, boxplots, curvas espectrais e mapas para examinar distribuição e heterogeneidade.

Analise assimetria, transformações possíveis, variabilidade dentro e entre talhões, regiões, safras e instrumentos. Examine correlações, colinearidade, relações lineares e não lineares, mas não interprete correlação como causalidade. Uma decisão de remover ou manter um outlier deve considerar o domínio e ser registrada.

### 4. Estatística descritiva e incerteza

Relate estimativas com unidades e medidas de dispersão, preferindo intervalos de confiança ou de previsão quando a inferência exigir. Diferencie:

- **variabilidade**: diferenças entre observações da população;
- **erro de medição**: incerteza introduzida pelo instrumento ou protocolo;
- **incerteza amostral**: variação decorrente das unidades observadas;
- **incerteza preditiva**: incerteza ao prever uma nova observação.

Use bootstrap, erros-padrão ou métodos apropriados ao desenho amostral quando forem necessários. Não apresente um intervalo como garantia fora da faixa, população ou condição ambiental usada para calculá-lo.

### 5. Comparações e relações entre variáveis

Escolha testes e estimadores de acordo com escala, distribuição, independência, tamanho amostral e desenho do estudo. Para comparar grupos,
considere diferenças de média ou mediana, tamanho de efeito e intervalos de confiança, não apenas o valor de `p`. Para relações contínuas, examine
correlação, regressão, resíduos, interações e possíveis confundidores.

Corrija ou controle comparações múltiplas quando várias hipóteses forem avaliadas. Em dados agrupados, use modelos hierárquicos, efeitos de grupo ou
erros-padrão robustos quando apropriado. Relate as limitações e não converta associação observacional em afirmação causal sem desenho causal adequado.

### 6. Preparação para modelagem

Divida os dados respeitando grupos, talhões, regiões ou tempo antes de ajustar transformações. Imputação, padronização, normalização, seleção de variáveis, PCA, cálculo de limiares e ajuste de hiperparâmetros devem usar somente o treino ou o fold de treinamento. O conjunto de teste deve permanecer intocado até a avaliação final.

Escolha regressão para alvos contínuos e classificação para categorias definidas operacionalmente. Use validação cruzada compatível com a estrutura
dos dados e compare modelos com os mesmos folds. Em dados espaciais e temporais, uma divisão aleatória pode gerar vazamento e desempenho
artificialmente alto.

### 7. Avaliação, diagnóstico e comunicação

Para regressão, avalie R², RMSE, MAE, viés, resíduos e incerteza. Para classificação, avalie acurácia, precisão, sensibilidade, F1, AUC-ROC, AUC-PR,
matriz de confusão e calibração quando necessário. Em problemas multiclasse, informe a média macro, micro ou ponderada.

Examine resíduos contra previsões e variáveis relevantes, erros por grupo, calibração, extrapolação, estabilidade e generalização. Apresente tabelas,
gráficos e mapas com unidades, escalas, datas, número de observações, intervalos de incerteza e limitações. Resultados estatísticos devem apoiar a
decisão agroambiental, sem substituir critérios agronômicos, validação de campo ou conhecimento do domínio.
